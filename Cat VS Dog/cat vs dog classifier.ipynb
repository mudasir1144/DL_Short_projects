```python
import tensorflow as tf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from tensorflow.keras import layers , Sequential
from tensorflow.keras.preprocessing.image import load_img , img_to_array
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    "animals",
    validation_split = 0.2,
    seed = 0,
    subset = "training",
    image_size = (128,128),
    batch_size = 32

)
```

```python
test_ds = tf.keras.utils.image_dataset_from_directory(
    "animals",
    validation_split = 0.2,
    seed = 0,
    subset = "validation",
    image_size = (128 , 128),
    batch_size = 32
)
```

```python
display(train_ds.element_spec)
```

```python
model = Sequential([
    layers.Conv2D(64, (3,3), activation = 'relu' , input_shape = (128 , 128 , 3)),
    layers.MaxPooling2D(2,2),
    layers.Conv2D(32, (3,3), activation = 'relu'),
    layers.MaxPooling2D(2,2),
    layers.Flatten(),
    layers.Dense(128 , activation = 'relu'),
    layers.Dense(1 , activation = 'sigmoid')
])
```

```python
model.compile(
    optimizer = 'adam',
    loss = 'binary_crossentropy',
    metrics = ['accuracy']
)
```

```python
model.fit(train_ds, validation_data = test_ds , epochs = 5)
```

```python
evalutaion = model.evaluate(test_ds)
print(evalutaion)
```

```python
image = load_img('animals/cat/00203-200124513.png', target_size=(128, 128))
input_arr = img_to_array(image)
input_array = np.array([input_arr])
predictions = model.predict(input_array)
if predictions > 0.5:
    print("The image given is a dog")
else:
    print("The image given is a cat")
```
